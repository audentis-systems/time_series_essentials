## User settings

```python
output_directory = 'output_test'
filename_show = '/home/emily/Desktop/projects/dj/song_recognition/data/full-goth-set.mp3'

path_library_parquet = '../database/music/output/playlist.parquet'  # not sure we are keeping this
soundfile_root_directory = '/media/emily/DJ_Backup/rekordbox_bak'  # not sure we are keeping this

sampling_rate = 22050
hop_length = 512 * 300
spark_memory = '70G'
octave_min_inclusive = 0
octave_max_inclusive = 8
percentile_cutoff = 0.8
approx_quantile_precision = 0.05
irq_min = 45.
irq_max = 60. * 11.

notebook_analyze_library = 'analyze-track-library.ipynb'
notebook_analyze_show = 'analyze-show-recording.ipynb'
notebook_join = 'join-aggregate-rank.ipynb'
```

## Load useful libraries

```python
import papermill as pm
import os
import shutil
```

## Create output directory (overwrite if it exists)

```python
if os.path.exists(output_directory) and os.path.isdir(output_directory):
    shutil.rmtree(output_directory)

os.mkdir(output_directory)
```

## Generate the track library features

```python
result = pm.execute_notebook(
   notebook_analyze_library,
   output_directory + '/' + notebook_analyze_library.replace('.ipynb', '-OUTPUT.ipynb'),
   parameters = {
        'output_directory' : output_directory,
        'path_library_parquet' : path_library_parquet,
        'sampling_rate' : sampling_rate,
        'hop_length' : hop_length,
        'spark_memory' : spark_memory,
        'octave_min_inclusive' : octave_min_inclusive,
        'octave_max_inclusive' : octave_max_inclusive,
        'soundfile_root_directory' : soundfile_root_directory,
   }
)
```

## Generate the show's feature vectors

```python
result = pm.execute_notebook(
   notebook_analyze_show,
   output_directory + '/' + notebook_analyze_show.replace('.ipynb', '-OUTPUT.ipynb'),
   parameters = {
        'output_directory' : output_directory,
        'filename_show' : filename_show,
        'sampling_rate' : sampling_rate,
        'hop_length' : hop_length,
        'spark_memory' : spark_memory,
        'octave_min_inclusive' : octave_min_inclusive,
        'octave_max_inclusive' : octave_max_inclusive,
    }
)
```
